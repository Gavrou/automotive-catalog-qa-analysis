```python
import pandas as pd
df = pd.read_csv('../data/raw/USA_Cars_Dataset.csv')
print("Columnas exactas del dataset:")
print(df.dtypes)
print("\nValores nulos por columna:")
print(df.isnull().sum())
print("\nFilas completamente duplicadas:", df.duplicated().sum())
df.head()
```

```python
print("Marcas únicas antes de limpiar:", sorted(df['Brand'].unique()))
print("\nCategorías de tamaño (Size):", sorted(df['Size'].unique()))

for col in ['Brand', 'Model', 'Size']:
    df[col] = df[col].str.strip()

df['Brand'] = df['Brand'].apply(lambda x: x if x.isupper() and len(x) <= 3 else x.title())

df_clean = df.rename(columns={
    'YEAR': 'model_year',
    'Brand': 'brand',
    'Model': 'model',
    'Size': 'vehicle_size',
    '(kW)': 'motor_power_kw',
    'CITY (kWh/100 km)': 'city_kwh_100km',
    'HWY (kWh/100 km)': 'hwy_kwh_100km',
    'COMB (kWh/100 km)': 'comb_kwh_100km',
    'CITY (Le/100 km)': 'city_le_100km',
    'HWY (Le/100 km)': 'hwy_le_100km',
    'COMB (Le/100 km)': 'comb_le_100km',
    'RATING': 'efficiency_rating',
    '(km)': 'range_km',
    'TIME (h)': 'charging_time_h'
})

df_clean['km_charged_per_hour'] = (df_clean['range_km'] / df_clean['charging_time_h']).round(1)
display(df_clean.head())
df_clean.to_csv('../data/processed/ev_catalog_clean.csv', index=False)
print("\nCatálogo limpio exportado con éxito a data/processed/ev_catalog_clean.csv.")
```
